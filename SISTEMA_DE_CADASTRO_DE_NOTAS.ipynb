```python
# Sistema de Gestão de Notas de Alunos

def cadastrar_notas():
    """Permite ao usuário inserir notas e armazená-las em uma lista."""
    notas = []
    qtd = int(input("Quantas notas deseja inserir? "))
    for i in range(qtd):
        nota = float(input(f"Digite a nota {i + 1}: "))
        notas.append(nota)
    return notas

def calcular_media(notas):
    """Calcula a média das notas da lista."""
    return sum(notas) / len(notas)

def determinar_situacao(media):
    """Retorna 'Aprovado' se média >= 7, senão 'Reprovado'."""
    if media >= 7:
        return "Aprovado"
    else:
        return "Reprovado"

def exibir_relatorio(notas, media, situacao):
    """Exibe o relatório final com notas, média e situação."""
    print("\n--- RELATÓRIO FINAL ---")
    print(f"Notas inseridas: {notas}")
    print(f"Média: {media:.2f}")
    print(f"Situação: {situacao}")

# Programa principal
notas_aluno = cadastrar_notas()
media_aluno = calcular_media(notas_aluno)
situacao_aluno = determinar_situacao(media_aluno)
exibir_relatorio(notas_aluno, media_aluno, situacao_aluno)
```
