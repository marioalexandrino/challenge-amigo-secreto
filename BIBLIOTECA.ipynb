```python
# ============================================
# SISTEMA DE GERENCIAMENTO DE LIVROS
# ============================================

# Importa a biblioteca Matplotlib para criação do gráfico
import matplotlib.pyplot as plt


# ============================================
# PASSO 1 - DEFINIÇÃO DA CLASSE LIVRO
# ============================================

class Livro:
    # Método construtor da classe
    def __init__(self, titulo, autor, genero, quantidade):
        self.titulo = titulo
        self.autor = autor
        self.genero = genero
        self.quantidade = quantidade

    # Método para apresentar os dados do livro
    def __str__(self):
        return (
            f"Título: {self.titulo} | "
            f"Autor: {self.autor} | "
            f"Gênero: {self.genero} | "
            f"Quantidade disponível: {self.quantidade}"
        )


# ============================================
# PASSO 2 - CRIAÇÃO DA LISTA DE LIVROS
# ============================================

# Lista vazia que armazenará os objetos Livro
livros = []


# ============================================
# PASSO 3 - FUNÇÃO PARA CADASTRAR UM LIVRO
# ============================================

def cadastrar_livro(titulo, autor, genero, quantidade):
    # Cria um novo objeto da classe Livro
    novo_livro = Livro(titulo, autor, genero, quantidade)

    # Adiciona o livro à lista
    livros.append(novo_livro)

    print(f'Livro "{titulo}" cadastrado com sucesso!')


# ============================================
# FUNÇÃO PARA LISTAR TODOS OS LIVROS
# ============================================

def listar_livros():
    # Verifica se a lista está vazia
    if not livros:
        print("Nenhum livro cadastrado.")
        return

    print("\n===== LIVROS CADASTRADOS =====")

    # Percorre todos os livros da lista
    for i, livro in enumerate(livros, start=1):
        print(f"{i}. {livro}")


# ============================================
# FUNÇÃO PARA BUSCAR LIVRO PELO TÍTULO
# ============================================

def buscar_livro(titulo):
    # Percorre a lista procurando pelo título
    for livro in livros:
        if livro.titulo.lower() == titulo.lower():
            print("\n===== LIVRO ENCONTRADO =====")
            print(livro)
            return livro

    print(f'\nLivro "{titulo}" não foi encontrado.')
    return None


# ============================================
# PASSO 4 - FUNÇÃO PARA GERAR O GRÁFICO
# ============================================

def gerar_grafico():
    # Dicionário para armazenar a quantidade de livros por gênero
    quantidade_generos = {}

    # Percorre os livros cadastrados
    for livro in livros:
        genero = livro.genero

        # Soma a quantidade disponível de cada gênero
        if genero in quantidade_generos:
            quantidade_generos[genero] += livro.quantidade
        else:
            quantidade_generos[genero] = livro.quantidade

    # Verifica se existem livros cadastrados
    if not quantidade_generos:
        print("Não existem livros para gerar o gráfico.")
        return

    # Cria o gráfico de barras
    plt.figure(figsize=(8, 5))

    plt.bar(
        quantidade_generos.keys(),
        quantidade_generos.values(),
        color="steelblue"
    )

    # Título e identificação dos eixos
    plt.title("Quantidade de Livros por Gênero")
    plt.xlabel("Gênero")
    plt.ylabel("Quantidade disponível")

    # Melhora a visualização dos nomes dos gêneros
    plt.xticks(rotation=30)

    # Exibe o gráfico
    plt.show()


# ============================================
# PASSO 5 - TESTE DO SISTEMA
# ============================================

# Cadastro de alguns livros para testar o sistema
cadastrar_livro(
    "Dom Casmurro",
    "Machado de Assis",
    "Romance",
    5
)

cadastrar_livro(
    "O Hobbit",
    "J. R. R. Tolkien",
    "Fantasia",
    4
)

cadastrar_livro(
    "1984",
    "George Orwell",
    "Ficção",
    3
)

cadastrar_livro(
    "Harry Potter e a Pedra Filosofal",
    "J. K. Rowling",
    "Fantasia",
    6
)

cadastrar_livro(
    "O Alienista",
    "Machado de Assis",
    "Romance",
    2
)


# ============================================
# TESTE 1 - LISTAR OS LIVROS
# ============================================

listar_livros()


# ============================================
# TESTE 2 - BUSCAR UM LIVRO
# ============================================

buscar_livro("O Hobbit")


# ============================================
# TESTE 3 - BUSCAR UM LIVRO INEXISTENTE
# ============================================

buscar_livro("Livro que não existe")


# ============================================
# TESTE 4 - GERAR O GRÁFICO
# ============================================

gerar_grafico()
```
